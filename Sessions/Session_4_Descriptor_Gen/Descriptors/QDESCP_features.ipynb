#### Organic molecule with SMILES input, CSEARCH performs conformational sampling with RDKit, QDESCP produces xtb properties


```python
import os
import shutil
import sys

# In Conda, these executables are in Library\bin in Windows and in bin in macOS/Linux.
bin_dir = os.path.join(
    sys.prefix,
    "Library",
    "bin"
) if os.name == "nt" else os.path.join(sys.prefix, "bin")

if os.path.isdir(bin_dir):
    os.environ["PATH"] = bin_dir + os.pathsep + os.environ.get("PATH", "")

obabel = shutil.which("obabel")
if obabel is None:
    raise RuntimeError(
        "Open Babel not found in the kernel environment. "
        "Check that it is installed in the selected Conda environment."
    )

print("Open Babel found:", obabel)
```

###### Step 1: CSEARCH to produce SDF files

```python
from aqme.qdescp import qdescp

# set working directory and CSV with SMILES strings
file = 'data/test.csv'

# run QDESCP descriptor generator, with:
# 1) Files to run (files=file.csv)
qdescp(input=file)
```
